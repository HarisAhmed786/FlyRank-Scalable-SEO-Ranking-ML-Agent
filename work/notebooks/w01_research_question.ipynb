This lane is useful because the dataset contains observed decline outcomes and allows us to investigate which signals are associated with pages needing attention.


## 2. The question: decision, action, cost of a wrong call

This work improves the decision of which page to refresh or audit next. SEO/content editors act on the output by reviewing the top-ranked pages first and choosing which content to update. A wrong recommendation wastes editor time on low-risk pages and may allow truly declining pages to keep losing traffic, so the cost is both wasted effort and missed recovery opportunity.

## 3. Quick look at the data (2-3 real numbers)

Over half of the sample pages (about 54.2%) are labeled as declining, so this is a common operational issue. The dataset has an average of about 5,200 search impressions over 90 days, meaning declines affect pages with real traffic. A simple signal like keyword search volume has almost no correlation with actual impressions, so ranking pages for refresh review requires more than one rule.

Another observation from exploration was that declining pages had a lower median content age (216 days) compared with stable and growing pages (around 292–300 days), suggesting that age alone is not a reliable refresh signal.
